# Household Data Visualization Notebook
This notebook contains the top 10 insightful data visualizations based on your dataset.

```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
from scipy.stats import f_oneway
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('DATASET/Family Income and Expenditure.csv')
```

## 1. Income vs Expenditure Composition

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

X = df[['Total Household Income']]
y = df['Total Food Expenditure']

model = LinearRegression()
model.fit(X, y)

# Predictions
pred = model.predict(X)

plt.figure()
plt.scatter(X, y, label='Actual', alpha=0.5)
plt.plot(X, pred, label='Predicted', linewidth=2)
plt.xlabel('Total Household Income')
plt.ylabel('Total Food Expenditure')
plt.title('Linear Regression: Income vs Food Expenditure')
plt.legend()
plt.show()
```

## 2. Regional Income & Food Spending

```python
X = df[['Total Food Expenditure', 'Total Household Income']]
y = df['Total Rice Expenditure']

model = LinearRegression()
model.fit(X, y)

pred = model.predict(X)

plt.figure()
plt.scatter(y, pred, alpha=0.5)
plt.xlabel('Actual Rice Expenditure')
plt.ylabel('Predicted Rice Expenditure')
plt.title('Multiple Linear Regression: Actual vs Predicted Rice Spending')
plt.show()


# df.groupby('Region')[['Total Household Income','Total Food Expenditure','Total Rice Expenditure']].mean().plot(kind='bar', figsize=(12,6))
# plt.title("Regional Income & Food Spending")
# plt.show()
```

## 3. Income vs Education & Employment

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

X = df[['Total Household Income', 'Total Food Expenditure']]
y = df['Agricultural Household indicator']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

model = LogisticRegression(max_iter=300)
model.fit(X_train, y_train)

pred = model.predict(X_test)

plt.figure()
plt.scatter(range(len(y_test)), y_test, label='Actual', alpha=0.6)
plt.scatter(range(len(pred)), pred, label='Predicted', alpha=0.6)
plt.title('Logistic Regression: Agricultural Household Prediction')
plt.ylabel('Indicator (0 or 1)')
plt.legend()
plt.show()

# plt.scatter(df['Household Head Highest Grade Completed'], df['Total Household Income'])
# plt.title("Income vs Education")
# plt.xlabel("Highest Grade Completed")
# plt.ylabel("Total Household Income")
# plt.show()
```

## 4. Housing Quality vs Income

```python
X = df[['Total Rice Expenditure', 'Fish and Seafood Expenditure']]
y = df['Meat Expenditure']

model = LinearRegression()
model.fit(X, y)

pred = model.predict(X)

plt.figure()
plt.scatter(y, pred, alpha=0.5)
plt.xlabel('Actual Meat Expenditure')
plt.ylabel('Predicted Meat Expenditure')
plt.title('Multiple Linear Regression: Meat Expense Prediction')
plt.show()

# plt.scatter(df['House Floor Area'], df['Total Household Income'])
# plt.title("Housing Quality vs Income")
# plt.xlabel("House Floor Area")
# plt.ylabel("Total Household Income")
# plt.show()
```

## 5. Appliance Ownership vs Wealth

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

X = df[['Total Household Income', 'Education Expenditure']]
y = (df['Number of Personal Computer'] > 0).astype(int)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

model = LogisticRegression(max_iter=300)
model.fit(X_train, y_train)

pred = model.predict(X_test)

plt.figure()
plt.scatter(range(len(y_test)), y_test, label='Actual', alpha=0.6)
plt.scatter(range(len(pred)), pred, label='Predicted', alpha=0.6)
plt.title('Logistic Regression: Predicting Home Computer Ownership')
plt.ylabel('Has Computer (0 or 1)')
plt.legend()
plt.show()

# appliances = ['Number of Television','Number of Refrigerator/Freezer','Number of Personal Computer','Number of Washing Machine','Number of Airconditioner']
# df.groupby(pd.qcut(df['Total Household Income'], 4))[appliances].mean().plot(kind='bar', figsize=(12,6))
# plt.title("Appliance Ownership vs Wealth")
# plt.show()
```

## 6. Family Composition vs Poverty Risk

```python
df['Total Non-Food Expenditure'] = df['Total Household Income'] - df['Total Food Expenditure']

X = df[['Total Food Expenditure', 'Total Rice Expenditure', 'Meat Expenditure']]
y = df['Total Non-Food Expenditure']

model = LinearRegression()
model.fit(X, y)

pred = model.predict(X)

plt.figure()
plt.scatter(y, pred, alpha=0.5)
plt.xlabel('Actual Non-Food Expenditure')
plt.ylabel('Predicted Non-Food Expenditure')
plt.title('Multiple Linear Regression: Predicting Non-Food Spending')
plt.show()

# plt.scatter(df['Total Number of Family members'], df['Total Household Income'])
# plt.title("Family Size vs Income")
# plt.xlabel("Family Members")
# plt.ylabel("Income")
# plt.show()
```
